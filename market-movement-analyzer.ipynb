```python
def stock_market_analyze(changes,strategies=abs):

  the_biggest_change_in_stock_price = max(changes,key=strategies)
  return the_biggest_change_in_stock_price

stock_price=[2.5, -4.1, 1.8, 6.3, -5.5, 0.4, -2.2]
print("The biggest change in stock price:",stock_market_analyze(stock_price))

def downside_move(x):
  return -x

print("The biggest downside movement in stock:",stock_market_analyze(stock_price,downside_move))
def upside_move(y):
  return y

print("The biggest upside movement in stock:",stock_market_analyze(stock_price,upside_move))



```
