```python
def trade_decision(stock_price, cash_balance, market_trend):
  if market_trend == True and stock_price < 50:
    decision = "Strong Buy"
  elif market_trend == False and cash_balance >= stock_price:
    decision = "Buy"
  else :
    decision = "Sell"
  return decision
trader_1 = trade_decision(40,50,1)
print(trader_1)
```
