```python
def evaluate_portfolio(cash, stock_price, shares_count):
  shares_count =  cash / stock_price
  if shares_count * stock_price > 5000 :
    answer = "Portfolio is safe and available for profits"
  else:
    answer = "Portfolio is in danger and recommend to add money"
  return answer

person_1 = evaluate_portfolio(30000,5,3)
print(person_1)
```
