```python
def calculate_portfolio_cost(ticker_symbols,is_premium_member):
  base_cost = 20 - 10 * is_premium_member
  cost_of_letter = len(ticker_symbols) * 2 /(is_premium_member + 1)
  cost = base_cost + cost_of_letter
  return cost
person_1 = calculate_portfolio_cost("ABCD",True)
print(person_1)
```
