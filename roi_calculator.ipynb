```python
def calculate_roi(buy_price,sell_price,shares):
  total_investment = buy_price * shares
  gross_return = sell_price * shares
  net_profit = gross_return - total_investment - 5.00
  # 5.00 is the commission fee.
  # ROI gives us the net profit percentage relative to the total investment.
  roi_percentage = (net_profit / total_investment) * 100
  return roi_percentage
x_roi = calculate_roi(1.20,1.80,150)
print("ROI Percentage = %",x_roi)
```
