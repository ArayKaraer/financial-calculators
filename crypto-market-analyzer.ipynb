```python
def analyze_crypto_prices(price):
  first_four_days_average = sum(price[:4]) / 4
  last_three_days_average = sum(price[-3:]) / 3
  max_price = max(price)
  min_price = min(price)
  high_prices = [price for price in price if price > 52]


  print("First four days average price:",first_four_days_average)
  print("Last three days average price:",last_three_days_average)
  print("Maximum price:",max_price)
  print("Minimum price:",min_price)
  print("Prices higher than 52:",high_prices)

crypto_prices = [45.5, 48.0, 47.2, 50.1, 52.3, 51.0, 54.5, 56.0, 55.2, 58.0]
analyze_crypto_prices(crypto_prices)






```
